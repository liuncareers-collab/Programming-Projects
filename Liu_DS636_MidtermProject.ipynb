<b>2. 5 Days Buy Low Rebalance Strategy</b>

```r
#New Function for Every 5 Days, should do a recursive function!
New_Day = 1
Interval = 5
New_Day_List = seq(1+Interval, 250, by = Interval)
Old_Day = 1
End_Day = "9-Jan-18"
Rebalance_Low = function(Close_Adj, Fund, Stocks, Portfolio, New_Day, Old_Day){
    #End_Day is when I want to stop 
    if (New_Day > which(rownames(Close_Adj) == End_Day)){
        return (Portfolio)
        }
    if (New_Day %in% New_Day_List){ #Every 5 days or interval
        #Checks to see which 'Adj Close' dropped most in percentage
        Adj_Close_Prices = select(Close_Adj, contains("Adj")); Adj_Close_Prices #includes only 'Adj_Close' columns
        Adj_Percentage = rename(arrange(melt(((Adj_Close_Prices[New_Day,] - Adj_Close_Prices[Old_Day-Interval,]) / (Adj_Close_Prices[Old_Day-Interval,])) * 100, id.vars = NULL), value), Stocks = variable, Percentage = value) #sorts dataframe by ascending percentage drops
    
        #Selling current shares
        Fund = sum(Portfolio[1:5] * Close_Adj[New_Day,Stocks]) + Portfolio$Cash; #takes previous shares, and sells them for current close value
        New_Stocks = sub("_Adj", "", Adj_Percentage$Stocks[1:5]) #gets name of stocks with lowest percent drops
        New_Close = Close_Adj %>% #creates data frame to select for only new stocks and close column
            select(contains(New_Stocks)) %>%
            select(contains("Close")) %>%
            slice(New_Day) #ensures only buying from value of current day

        #Buying new stock and creating new portfolio
        Stocks = colnames(New_Close)
        Shares = floor((Fund / 5) / New_Close)
        Cash = sum((((Fund / 5) / New_Close) - Shares) * New_Close)
        MTM = Cash + sum(Shares * New_Close)
        Portfolio = data.frame(Shares, Cash, MTM)
        Rebalance_Low(Close_Adj, Fund, Stocks, Portfolio, New_Day + 1, Old_Day + 1)
        }
    else{ #MTM will update every day
        MTM = Portfolio$Cash + sum(Portfolio[1:5] * Close_Adj[New_Day, Stocks])
        Rebalance_Low(Close_Adj, Fund, Stocks, Portfolio, New_Day + 1, Old_Day + 1)
        }
    }

Portfolio = Rebalance_Low(Close_Adj, Fund, Stocks, Portfolio, New_Day, Old_Day); Portfolio

```

<b>3: Adding Dividends</b>

```r
#Recursive function too intensive, moving to for loop
r1 = 1
r2 = 2
c1 = 1
c2 = 2
Row_Names = c()
Dates = data.frame()
Dividends = data.frame()
Dividend = function(r1, r2, c1, c2, Dividends){
    if ((r2 >= nrow(Close_Adj)) & (c2 >= ncol(Close_Adj))){
        colnames(Dividends) = "Dividends"
        return (Dividends)
    } else if ((r2 >= nrow(Close_Adj)) & (c2 <= ncol(Close_Adj))){
        return (Dividend(r1 = 1, r2 = 2, c1+2, c2+2, Dividends))
    } else{
        Ratio = ((Close_Adj[r1,c1] / Close_Adj[r2,c1]) - (Close_Adj[r1,c2] / Close_Adj[r2,c2])) * Close_Adj[r2,c1]
        if (Ratio > 0.1){
            Row_Names <<- c(Row_Names, colnames(Close_Adj)[c2])
            Dates = c(Dates, rownames(Close_Adj[r2,]))
            Dividends = rbind(Dividends, Ratio)
            }
        return(Dividend(r1+1, r2+1, c1, c2, Dividends))
    }
    }
Dividend(r1, r2, c1, c2, Dividends)
print(Row_Names)
```

```r
#For Loop

#initializing  variables 
Dividends = data.frame()
Stock_Names = data.frame()
Dates = data.frame()

for (c2 in seq(2, 20, by = 2)){ #will run through adj_close column 2:20 (for each of the 10 stocks)
    for (r2 in seq(2, 250, by = 1)){ #for each stock, run through all dates
        if ((((Close_Adj[r2-1,c2-1] / Close_Adj[r2,c2-1]) - (Close_Adj[r2-1,c2] / Close_Adj[r2,c2])) * Close_Adj[r2,c2-1]) > 0.1){ #if dividend ratio surpasses threshold (0.1)
            Dividends = rbind(Dividends, ((Close_Adj[r2-1,c2-1] / Close_Adj[r2,c2-1]) - (Close_Adj[r2-1,c2] / Close_Adj[r2,c2])) * Close_Adj[r2,c2-1]) #add dividend to data frame
            Stock_Names = rbind(Stock_Names, colnames(Close_Adj)[c2-1]) #Add stock name to data frame
            Dates = rbind(Dates, rownames(Close_Adj[r2,])) #Add dates to data frame 
            }
        }
    }
Dividend_df = data.frame(Dividends, Stock_Names, Dates) #combine into one data frame
colnames(Dividend_df) = c("Dividends", "Stocks", "Dates") #give column names to data frame
Dividend_df

#Sets us up for using these dividends for our Rebalance function
```

```r
#Initializes portfolio so we don't have to go back everytime
Fund = 5000000 #total money in fund
Stocks = c("IBM_Close", "MSFT_Close", "GOOG_Close", "AAPL_Close", "AMZN_Close") #represents the close column of the stocks we want to buy
Days = "2-Jan-18"
Investment = function(Fund, Close_Adj, Stocks, Days){
    Shares = floor((Fund / 5) / Close_Adj[Days, Stocks])
    Cash = sum((((Fund / 5) / Close_Adj[Days, Stocks]) - Shares) * Close_Adj[Days, Stocks])
    MTM = Cash + sum(Shares * Close_Adj[Days, Stocks])
    Portfolio = data.frame(Shares, Cash, MTM)
    }

Portfolio = Investment(Fund, Close_Adj, Stocks, Days)

New_Day = 1
Old_Day = 1
End_Day = "28-Dec-18"
Dividend_Cash = 0
Interval = 5
New_Day_List = seq(1+Interval, 250, by = Interval)
MTM_Low_List = c() #For use in part 7, daily MTMs added here

#Redid Rebalance function to include dividend calculations
Rebalance_Low_Dividend = function(Close_Adj, Fund, Stocks, Portfolio, New_Day, Old_Day, Dividend_Cash){
    if (New_Day > which(rownames(Close_Adj) == End_Day)){
        return (Portfolio)
        }

    #Dividend Calculations
    if (rownames(Close_Adj[New_Day,]) %in% Dividend_df$Dates){ #If the current day lands on a dividend day
        if (Dividend_df$Stocks[(match(rownames(Close_Adj[New_Day,]), Dividend_df$Dates))] %in% Stocks){ #looks to see if we are invested in current dividend stock
            Dividend_Cash = Dividend_df$Dividends[(match(rownames(Close_Adj[New_Day,]), Dividend_df$Dates))] * Portfolio[,match(Dividend_df$Stocks[(match(rownames(Close_Adj[New_Day,]), Dividend_df$Dates))], Stocks)] #dividend * shares
        }
        }
    if (New_Day %in% New_Day_List){ 
        Adj_Close_Prices = select(Close_Adj, contains("Adj")); Adj_Close_Prices 
        Adj_Percentage = rename(arrange(melt(((Adj_Close_Prices[New_Day,] - Adj_Close_Prices[Old_Day-Interval,]) / (Adj_Close_Prices[Old_Day-Interval,])) * 100, id.vars = NULL), value), Stocks = variable, Percentage = value)

        Fund = sum(Portfolio[1:5] * Close_Adj[New_Day,Stocks]) + Portfolio$Cash;
        New_Stocks = sub("_Adj", "", Adj_Percentage$Stocks[1:5])
        New_Close = Close_Adj %>%
            select(contains(New_Stocks)) %>%
            select(contains("Close")) %>%
            slice(New_Day)

        Stocks = colnames(New_Close)
        Shares = floor((Fund / 5) / New_Close)
        Cash = sum((((Fund / 5) / New_Close) - Shares) * New_Close) + Dividend_Cash #Adds dividend to cash account
        MTM = Cash + sum(Shares * New_Close)
        MTM_Low_List <<- c(MTM_Low_List, MTM) #For part 7
        Portfolio = data.frame(Shares, Cash, MTM)
        Rebalance_Low_Dividend(Close_Adj, Fund, Stocks, Portfolio, New_Day + 1, Old_Day + 1, Dividend_Cash = 0) #Resets dividend, so it is not added again next recursion
        }    
    else{
        MTM = Portfolio$Cash + sum(Portfolio[1:5] * Close_Adj[New_Day, Stocks])
        MTM_Low_List <<- c(MTM_Low_List, MTM)
        Rebalance_Low_Dividend(Close_Adj, Fund, Stocks, Portfolio, New_Day + 1, Old_Day + 1, Dividend_Cash = 0)
        } 
    }

Portfolio = Rebalance_Low_Dividend(Close_Adj, Fund, Stocks, Portfolio, New_Day, Old_Day, Dividend_Cash); Portfolio
```
