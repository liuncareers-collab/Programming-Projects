```python
#Initializing Queue
start_state = [False, False, False, False]
goal_state = [True, True, True, True]
Queue = [start_state]
visited = [start_state]
#For output indexing
elements = ['Man','Lion','Goat','Grass']
starting_side = []
target_side = []
shortest_path = [[start_state]] #shortest path will always start with start state
```

```python
def ValidationFunction(current_state):
    if current_state[1] == current_state[2] and current_state[0] != current_state[1]: #lion and goat w/o man
      return False
    if current_state[2] == current_state[3] and current_state[0] != current_state[2]: #goat and grass w/o man
      return False
    else:
      return True
```

```python
def GenerationFunction(current_state):
  generated_states = [] #list container for generated states
  for index in range(0, len(current_state)):
    if index == 0:
      temp_state = current_state.copy() #creates a temporary copy of the current state
      temp_state[0] = not temp_state[0] #switches state of man
      generated_states.append(temp_state)
    else:
      if current_state[index] == current_state[0]:
        temp_state = current_state.copy()
        temp_state[0] = not temp_state[0]
        temp_state[index] = not temp_state[index] #switches state of other elements
        generated_states.append(temp_state)
  return generated_states
```

```python
while Queue:
  if len(shortest_path) == 1: #initialize output, as we'll always start w/ this configuration
      print("starting side: ['Man', 'Lion', 'Goat', 'Grass']")
      print("target side: []")
      print('')

  #clears these variables for new output
  starting_side.clear()
  target_side.clear()

  #makes current state the first in queue
  current_state = Queue.pop(-0)

  #checks if we have the finished state
  if current_state == goal_state:
    break

  generated_states = GenerationFunction(current_state) #generates child states
  for state in generated_states:
    if state not in visited :
      validated = ValidationFunction(state) #validates states
      if validated == True:
        visited.append(state) #adds to visited variable
        if len(shortest_path) > 1: #checks for any same-level sequences
          if state[0] != shortest_path[-1][0]: #makes sure we are in correct sequence (man should not have same state twice in sequence)
            shortest_path.append(state)

            #output for sequence will only be done when shortest path is added to
            for index in range(len(shortest_path[-1])):
              if shortest_path[-1][index] == False:
                starting_side.append(elements[index])
              elif shortest_path[-1][index] == True:
                target_side.append(elements[index])
            print('starting side: ', starting_side)
            print('target side: ', target_side)
            print('')

        elif len(shortest_path) <= 1: #beginning of sequence.
          shortest_path.append(state)

          for index in range(len(shortest_path[-1])):
            if shortest_path[-1][index] == False:
              starting_side.append(elements[index])
            elif shortest_path[-1][index] == True:
              target_side.append(elements[index])
          print('starting side: ', starting_side)
          print('target side: ', target_side)
          print('')

        Queue.append(state) #adds to queue variable
```
