```python
#initializing variables
visited = []
start_state = [False, False, False, False]
goal_state = [True, True, True, True]
names = ['Man', 'Lion', 'Goat', 'Grass']
```

```python
#checks if states are valid
def validationFunction(current_state):
    if current_state[1] == current_state[2] and current_state[0] != current_state[1]: #lion and goat w/o man
      return False
    if current_state[2] == current_state[3] and current_state[0] != current_state[2]: #goat and grass w/o man
      return False
    else:
      return True
```

```python
#child node generation
def stateGeneration(current_state):
  generated_states = []
  for position in range(0,len(current_state)):
    temp_state = current_state.copy() #so we don't change current state while generating states
    if position == 0: #man only
      temp_state[0] = not temp_state[0]
      generated_states.append(temp_state)
    elif current_state[position] == current_state[0]: #man + ?
      temp_state[0] = not temp_state[0]
      temp_state[position] = not temp_state[position]
      generated_states.append(temp_state)
  return generated_states
```

```python
def dfsAlgo(current_state, visited):

  #for printing path
  starting_side = []
  target_side = []

  #goal condition
  if current_state == goal_state:
    # printing path
    for index in range(0, len(current_state)):
      if current_state[index] == False:
        starting_side.append(names[index])
      if current_state[index] == True:
        target_side.append(names[index])
    print('starting side: ', starting_side)
    print('target side: ', target_side)

    return True

  if not validationFunction(current_state): #checks to see if function is valid
    return False #will return back to later for loop iteration, moving onto next child

  if current_state in visited:
    return False

  visited.append(current_state)

  #printing path
  for index in range(0, len(current_state)):
    if current_state[index] == False:
      starting_side.append(names[index])
    if current_state[index] == True:
      target_side.append(names[index])
  print('starting side: ', starting_side)
  print('target side: ', target_side)
  print('')

  #node generation
  for child in stateGeneration(current_state):
    if dfsAlgo(child, visited): #recursive function
      return 'Complete' #if the function returns True - which only happens when we reach the goal - exit function.

  return False #if all the children return False, then we have no more stack. End function.
```

```python
dfsAlgo(start_state, visited) #starts algorithm
```
